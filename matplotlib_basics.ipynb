```python
fig=plt.figure()

axes1 = fig.add_axes([0,0,1,1])
axes1.plot(a,b)
axes1.set_xlim(0,8)
axes1.set_ylim(0,8000)
axes1.set_xlabel('A')
axes1.set_ylabel('B')
axes1.set_title('Power of 4')

axes2 = fig.add_axes([0.2,0.5,0.25,0.25])
axes2.plot(x,y)
axes2.set_xlim(1,2)
axes2.set_ylim(0,25)
axes2.set_xlabel('A')
axes2.set_ylabel('B')
axes2.set_title('Zoomed in')
```

```python
fig=plt.figure(figsize=(2,2),dpi=200)
axes1=fig.add_axes([0,0,1,1])
axes1.plot(a,b)

fig.savefig('new_figure.png',bbox_inches='tight')
```

```python
fig,axes=plt.subplots(nrows=3,ncols=1)

for ax in axes:
    ax.plot(x,y)

plt.tight_layout()
```

```python
fig,axes=plt.subplots(nrows=2,ncols=2,figsize=(4,10),dpi=200)
axes[0][0].plot(x,y)
axes[1][0].plot(a,b)
axes[1][0].set_ylabel('Y label 1,0')
axes[1][0].set_xlim(2,6)
axes[0][1].plot(x,y)
axes[1][1].plot(a,b)
axes[1][1].set_title('Title 1,1')
#plt.tight_layout()
fig.suptitle("Figure level", fontsize=16)
fig.subplots_adjust(wspace=1,hspace=0.5)
#fig.set_figwidth(10)

fig.savefig('new_subplots.png',bbox_inches='tight')
```

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
ax.plot(x,x,label="X vs X")
ax.plot(x,x**2,label="X vs X^2")
#ax.legend(loc=4) 0 is default
ax.legend(loc=(1.1,0.5))
```

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
#ax.plot(x,x,color='purple')
ax.plot(x,x,color='#650dbd') #RGB HEX code from google
ax.plot(x,x+1,color="yellow")
```

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
#ax.plot(x,x,color='purple')
ax.plot(x,x,color='#650dbd',lw=10) #RGB HEX code from google
ax.plot(x,x+1,color="yellow",linewidth=1.5)
```

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
#ax.plot(x,x,color='purple')
ax.plot(x,x,color='#650dbd',lw=6,ls='--') #RGB HEX code from google
ax.plot(x,x+1,color="green",linewidth=1.5,ls='-.')

#possible linestyle options  '-', '--', '-.', ':', 'None', ' ', '', 'solid', 'dashed', 'dashdot', 'dotted'
```

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
lines  = ax.plot(x,x,color='purple',lw=6,ls='--')
lines[0].set_dashes([1,2,1,2,10,2])
```

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
lines  = ax.plot(x,x,color='purple',lw=3,marker='o',ls='--')
```
