```python
d = {'apple':5, 'banana':4, 'orange':4}
```

```python
d_disc = {'apple':0.5, 'banana': 1}
```

```python
total =0
for key, value in d.items():
    for key1, value1 in d_disc.items():
        if key == key1:
            total += value - value1

print(total)
```

```python
total = 0
for key, value in zip(d.keys(),d.values()):
    for key1, value1 in zip(d_disc.keys(), d_disc.values()):
        if key == key1:
            total += value - value1
print(total)
```

```python
total =0
for d_key in d.keys():
    for d_disc_key in d_disc.keys():
        if d_key == d_disc_key:
            total += d[d_key] - d_disc[d_disc_key]
print(total)
```

```python
total =0
for key, value in d.items():
    if key in d_disc:
        total+= value-d_disc[key]
print(total)
```

```python
print(sum(d[key] - d_disc[key] for key in d if key in d_disc))
```

One for loop and one if loop is O(n)

2 for loops is O(n*m)
