**Method 4 using random distribution**

```python
np.random.rand(5) #uniform distribution on number between 0 and 1, for the provided count
```

```python
np.random.rand(4,2) #n dimensional uniform distribution
```

```python
np.random.randn(5) # standard uniform distribution allow negative numbers, mean 0 and variance 1
```

```python
np.random.randn(3,3)
```

```python
np.random.randint(10,20,4) # integers between start and stop, for the shape of array
```

```python
np.random.randint(10,30,(3,3))
```

**with seed same random numbers everytime, when used in same cell**
**without seed rand provides, different array**
**any number can be used in seed, not only 100(that's example)**

```python
np.random.seed(100)
np.random.rand(3)
```

```python
np.random.seed(100)
np.random.rand(3)
```
