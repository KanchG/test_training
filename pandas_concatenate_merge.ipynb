#### Concatenate

```python
import numpy as np
import pandas as pd
```

```python
data_one = {'A':['A0','A1','A2','A3'],'B':['B0','B1','B2','B3']}
```

```python
data_two = {'C':['C0','C1','C2','C3'],'D':['D0','D1','D2','D3']}
```

```python
one = pd.DataFrame(data_one)
```

```python
two = pd.DataFrame(data_two)
```

```python
one
```

```python
two
```

```python
pd.concat([one,two],axis =1) # concat along columns
```

```python
pd.concat([one,two],axis=0) #concat along rows
```

```python
one
```

```python
two.columns = one.columns
```

```python
two
```

```python
mydf = pd.concat([one,two],axis=0)
```

```python
mydf
```

```python
mydf.index = range(len(mydf))
```

```python
mydf
```

#### Merge (Inner,Outer, Left/Right)

```python
registrations = pd.DataFrame({'reg_id':[1,2,3,4],'name':['Andrew','Bobo','Claire','David']})
logins = pd.DataFrame({'log_id':[1,2,3,4],'name':['Xavier','Andrew','Yolanda','Bobo']})
```

```python
registrations
```

```python
logins
```

***INNER MERGE***

```python
pd.merge(registrations,logins,how='inner',on='name')
```

```python
pd.merge(logins,registrations,how='inner',on='name') #order doesnt matter for inner merge
```

***LEFT MERGE***

```python
pd.merge(registrations,logins,how='left',on='name')
```

***RIGHT MERGE***

```python
pd.merge(registrations,logins,how='right',on='name')
```

***OUTER MERGE***

```python
pd.merge(registrations,logins,how='outer',on='name') #order of tables doesnt matter as inner merge
```

```python
registrations = registrations.set_index('name')
```

```python
registrations
```

```python
logins
```

```python
pd.merge(registrations,logins,left_index=True,right_on='name',how='inner')
```

```python
registrations = registrations.reset_index()
```

```python
registrations
```

```python
registrations.columns=['reg_name','reg_id']
```

```python
registrations
```

```python
results = pd.merge(registrations,logins,how='inner',left_on='reg_name',right_on='name')
```

```python
results
```

```python
results.drop('reg_name',axis=1)
```
