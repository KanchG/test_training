## Distribution plots
***Rug plot - Histogram - KDE (Kernel Density Estimation)***

```python
sns.rugplot(x='salary',data=df,height=0.5)
```

```python
sns.set(style = 'darkgrid') # 'whitegrid'or 'white'or 'dark'or 'ticks'
sns.displot(data=df,x='salary',bins =30,
          #  color='red',edgecolor='blue',
          #  linewidth=3,ls='--',
            rug=True, kde=True)
```

```python
sns.histplot(data=df,x='salary')
```

```python
sns.kdeplot(data=df,x='salary')
```

```python
np.random.seed(42)
sample_ages=np.random.randint(0,100,200)
```

```python
sns.kdeplot(data=sample_ages,x='ages',clip=[0,100],bw_adjust=0.1,fill=True)
```

## Categorial plots

```python
df['division'].value_counts()
```

```python
plt.figure(figsize=(10,4))
sns.countplot(data=df,x='division',hue='level of education',palette='Set2')
plt
```

```python
plt.figure(figsize=(10,4))
sns.barplot(data=df,x='level of education',y='salary',estimator=np.mean,errorbar='sd',hue='division')
plt.legend(bbox_to_anchor=(1.05,1))
```

```python
df=pd.read_csv('StudentsPerformance.csv')
```

```python
df.head()
```

```python
plt.figure(figsize=(10,4))
sns.boxplot(data=df,y='math score',x='parental level of education',hue='test preparation course')
plt.legend(bbox_to_anchor=(1.05,0.5))
```

```python
sns.violinplot(data=df,y='math score',x='parental level of education',hue='test preparation course',split=True,inner=None)
```

```python
sns.swarmplot(data=df,x='math score',size=2,y='gender',hue='test preparation course', dodge=True)
```

```python
sns.boxenplot(x='math score',y='gender',data=df,hue='test preparation course')
```

## Comparison plots
***JOINT PLOT and PAIR PLOT***

```python
df.head()
```

```python
sns.jointplot(data=df,x='math score',y='reading score',hue="gender") #scatter kind or hist kind or kde 
```

```python
sns.pairplot(data=df,hue='gender',corner=True) 
```

## GRID SYSTEM

```python
sns.catplot(data=df,x='gender',y='math score',kind='box',row='lunch',col='test preparation course')
```

```python
g = sns.PairGrid(df,hue='gender')
g=g.map_upper(sns.scatterplot)
g=g.map_diag(sns.kdeplot)
g=g.map_lower(sns.kdeplot)
g=g.add_legend()
```
