#### Conditional Filtering

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('tips.csv')
```

```python
df.head()
```

```python
bool_series = df['total_bill'] > 40
```

```python
df[bool_series] # no need of variable
```

```python
df[df['total_bill'] > 40]
```

```python
df[df['sex'] == 'Male']
```

**Multiple Conditional filtering**

```python
# AND & ----- Both condn need to be true

# OR | ----- Either condn is true
```

```python
df[(df['total_bill'] > 40) & (df['sex'] == "Male")]
```

```python
df[(df['total_bill'] > 40) | (df['sex'] == "Male")]
```

```python
#df[(df['day'] =='Sun') | (df['day'] == 'Sat') | (df['day'] == 'Fri')]
```

```python
options = ['Sat','Sun']
```

```python
df[df['day'].isin(options)]

#can be used directly as below too

#df[df['day'].isin(['Sat','Sun'])]
```
