#### USEFUL METHODS - APPLY - SINGLE COLUMN

```python
def last_four(num):
    return str(num)[-4:]
```

**these apply function can return only one single value**

```python
df['last_four'] = df['CC Number'].apply(last_four)
```

```python
df['total_bill'].mean()
```

```python
def yelp(price):
    if price < 10:
        return '$'
    elif price >= 10 and price < 30:
        return '$$'
    else:
        return '$$$'
```

```python
df['yelp'] = df ['total_bill'].apply(yelp)
```

```python
df.head()
```

#### USEFUL METHODS - APPLY - MULTIPLE COLUMN

```python
df['total_bill'].apply(lambda num: num*2)   
```

```python
df.head()
```

```python
def quality(total_bill,tip):
    if tip/total_bill > 0.25:
        return "Generous"
    else:
        return "Other"
```

```python
df['Quality'] =df[['total_bill','tip']].apply(lambda df: quality(df['total_bill'],df['tip']),axis=1)
```

```python
df.head()
```

```python
df['Quality'] = np.vectorize(quality)(df['total_bill'],df['tip'])
```

```python
df.head()
```
